```python
print("Billing System")
Item1=input("Enter the Fisrt Item name: ")
Item2=input("Enter the Second Item name: ")

price1=input("Enter the price of the First item: ")
price2=input("Enter the price of the Second item: ")

Quantity_item1=input("Enter the Quantity of Item1: ")
Quantity_item2=input("Enter the Quantity of Item2: ")

discount1=input("Enter the discount of the First item: ")
discount2=input("Enter the discount of the Second item: ")

price1 = int(price1)
price2= int(price2)
Quantity1=int(Quantity_item1)
Quantity2=int(Quantity_item2)

discount1=int(discount1)
discount2=int(discount2)
discount=discount1+discount2

total_price= (price1*Quantity1 - discount1*Quantity1) + (price2*Quantity2 -discount2*Quantity2)

final_price=total_price-discount
print("final price is:",final_price)
```
